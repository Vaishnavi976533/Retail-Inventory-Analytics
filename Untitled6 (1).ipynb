```python
category_count = df['Category'].value_counts()

print(category_count)
```

```python
category_sales = df.groupby('Category')['Units Sold'].sum()

print(category_sales)
```

```python
top_products = df.groupby('Product Name')['Units Sold'].sum()

top_products = top_products.sort_values(
    ascending=False
).head(10)

print(top_products)
```

```python
low_stock = df.sort_values(
    by='Closing Stock'
).head(10)

print(low_stock[['Product Name',
                 'Closing Stock']])
```

```python
supplier_inventory = df['Supplier Name'].value_counts()

print(supplier_inventory)
```

```python
category_sales.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Category Wise Units Sold')
plt.xlabel('Category')
plt.ylabel('Units Sold')

plt.show()
```

```python
df['Category'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Category Distribution')

plt.show()
```

```python
plt.hist(
    df['Units Sold'],
    bins=20
)

plt.title('Units Sold Distribution')
plt.xlabel('Units Sold')
plt.ylabel('Frequency')

plt.show()
```

```python
top10 = df.groupby('Product Name')[
    'Units Sold'
].sum().sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10,5))

plt.plot(
    top10.index,
    top10.values
)

plt.xticks(rotation=90)

plt.title(
    'Top Selling Products'
)

plt.show()
```

```python
import seaborn as sns

corr = df[
[
'Opening Stock',
'Units Sold',
'Closing Stock',
'Reorder Point'
]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title(
'Correlation Heatmap'
)

plt.show()
```
